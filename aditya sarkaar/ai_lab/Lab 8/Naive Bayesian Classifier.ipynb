# Bayesian Classifier

```python
# Import LabelEncoder
from sklearn import preprocessing
#for data manipulation
import pandas as pd
#naive bayesian classifier model
from sklearn.naive_bayes import GaussianNB 
#for data operation
import numpy as np
```

```python
golf = pd.read_csv("golf.csv")
golf = golf.drop(["Unnamed: 5"] ,axis = 1)
golf
```

```python
le = preprocessing.LabelEncoder()
# Converting string labels into numbers.
outlook = le.fit_transform(golf["OUTLOOK"]).reshape(-1, 1)
temperature = le.fit_transform(golf["TEMPERATURE"]).reshape(-1, 1)
humidity = le.fit_transform(golf["HUMIDITY"]).reshape(-1, 1)
windy = le.fit_transform(golf["WINDY"]).reshape(-1, 1)
play_golf = le.fit_transform(golf["PLAY GOLF"]).reshape(-1, 1)

#Combinig different features into single listof tuples
#features = np.concatenate((outlook, temperature, humidity, windy), axis = 1)
#features = np.concatenate((outlook, temperature.T), axis = 0)
features_pd = pd.DataFrame(outlook, columns = ['outlook'])
features_pd['temperature'] = temperature
features_pd['humidity'] = humidity
features_pd['windy'] = windy
print(features_pd)
features = features_pd.to_numpy()
features
```

```python
#Create a Gaussian Classifier
model = GaussianNB()

features.reshape(-1, 2)
# Train the model using the training sets
model.fit(features,play_golf)

#Predict Output
predicted= model.predict([[2,0,1,0]]) # sunny, hot, normal, false
print("Predicted Value:", predicted)
if predicted == 1:
    print("Golf play : Yes")
else:
     print("Golf play : No")   
```
