```python
# %pip install scikit-learn pandas matplotlib seaborn

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()
X, y = iris.data, iris.target
```

```python
# First split: keep 20% as a final test set
X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
```

```python
# Second split: create training and validation sets
X_train, X_val, y_train, y_val = train_test_split(
X_temp, y_temp, test_size=0.25, random_state=42, stratify=y_temp
)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)
```

```python
print("Training samples:", len(X_train))
print("Validation samples:", len(X_val))
print("Test samples:", len(X_test))
```

The test set should not be used while choosing a model or hyperparameter because doing so lets the model “see” the final evaluation data during selection, which can make the results look better than they really are. Instead, the validation set is used to compare models and tune parameters, while the test set is kept for a final, unbiased evaluation.
